```python
import pandas as pd
import scipy.stats as stats
from sklearn.linear_model import LinearRegression
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('advanced_cleaned_healthcare.csv')

# 1. Statistical Hypothesis Testing[cite: 2, 3]
# Is there a significant difference in billing between Emergency and Elective?
group1 = df[df['Admission_Type'] == 'Emergency']['Billing_Amount']
group2 = df[df['Admission_Type'] == 'Elective']['Billing_Amount']
t_stat, p_val = stats.ttest_ind(group1, group2)
print(f"P-Value for Billing Difference: {p_val:.4f}") # If < 0.05, it's significant

# 2. Predictive Analysis (The "Complex" Part)[cite: 2]
# Predicting Billing Amount based on Age and Risk Score
X = df[['Age', 'Patient_Risk_Score']]
y = df['Billing_Amount']
model = LinearRegression().fit(X, y)
print(f"Model R-squared: {model.score(X, y):.4f}")

# 3. Advanced Visualization: Correlation Heatmap
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Patient Health Metric Correlations')
plt.show()
```
